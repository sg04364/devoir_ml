```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('xxx/Data.csv', sep = ';')
```

```python
def gradient(pays, lr):
    """
    param :
    pays: str
    lr: float
    ------------
    return :
    a: float
    b: float
    """
    X = df[pays].values[:-1]
    y = df[pays].values[1:]
    n = len(X)

    eps = 10e-6

    a = 1
    b = 1
    cout_i1 = 1
    cout_i = 0
    grad_a = 1
    grad_b = 1

    while abs(cout_i1 - cout_i) > eps or abs(grad_a) > eps or abs(grad_b) > eps:

        cout_i = np.sum((a*X + b - y)**2) / (2 * n)

        grad_a = np.sum((a*X + b - y) * X) / n
        grad_b = np.sum((a*X + b - y)) / n

        a = a - lr * grad_a
        b = b - lr * grad_b

        cout_i1 = np.sum((a*X + b - y)**2) / (2 * n)

    a = round(a,4)
    b = round(b,4)

    return a, b
```

```python
a, b = gradient('USA', 0.01)
print(a, b)

plt.plot(df.index, df['USA'], color='red', label='Données réelles')
x = df['USA'].values[:-1]
plt.plot(df.index[:-1], a * x + b, color='blue', label='Prédictions')
plt.title("USA")
plt.xlabel("Temps")
plt.ylabel("Valeur")
plt.legend()
plt.show()
```
